# Customer Churn Prediction and Analysis

## 1. Project Introduction

Customer churn means when a customer stops using a company's products or services. Customer churn is an important problem for telecommunication companies because losing customers can reduce business revenue.

In this project, I used Python to analyse customer data and identify the important factors related to customer churn. I also applied Machine Learning techniques to predict customer churn and group customers into different segments.

## 2. Problem Statement

Telecommunication companies face a problem when customers stop using their services. This is known as customer churn. Losing customers can affect the company's revenue and growth.

The main problem addressed in this project is to understand why customers leave a telecommunication company and identify customers who may be likely to leave. Data analysis and Machine Learning can help companies understand customer behaviour and take steps to improve customer retention.

````python
features = [
    'Gender',
    'Age',
    'Married',
    'Number of Dependents',
    'Tenure Months',
    'Phone Service',
    'Multiple Lines',
    'Internet Service',
    'Internet Type',
    'Online Security',
    'Online Backup',
    'Device Protection Plan',
    'Premium Tech Support',
    'Streaming TV',
    'Streaming Movies',
    'Streaming Music',
    'Unlimited Data',
    'Contract',
    'Paperless Billing',
    'Payment Method',
    'Monthly Charges',
    'Total Charges'
]

# Keep only columns available in the dataset
features = [col for col in features if col in df_clean.columns]

print("Selected Features:")
print(features)

print("Total Features:", len(features))
````

````python
X = df_clean[features].copy()

y = df_clean['Churn Label'].astype(str).str.strip().map({
    'Yes': 1,
    'No': 0
})

print("Target Variable:")
print(y.value_counts(dropna=False))
````

````python
# Separate numerical and categorical columns

numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

print("Numerical Features:", numeric_features)
print("Categorical Features:", categorical_features)
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
````

````python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
````

````python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Training Data Shape:", X_train_processed.shape)
print("Testing Data Shape:", X_test_processed.shape)
````

````python
from sklearn.linear_model import LogisticRegression

print("Logistic Regression imported successfully!")
````

````python
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print("Model created successfully!")
````

````python
model.fit(X_train_processed, y_train)

print("Model training completed successfully!")
````

````python
y_pred = model.predict(X_test_processed)

print("Prediction completed successfully!")

print("First 10 Predictions:")
print(y_pred[:10])
````

````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
````

````python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=['No Churn', 'Churn']
))
````

````python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No Churn', 'Churn'],
    yticklabels=['No Churn', 'Churn']
)

plt.title('Confusion Matrix')
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')

plt.show()
````

````python
from sklearn.cluster import KMeans

print("K-Means imported successfully!")
````

````python
X_cluster = preprocessor.transform(X)

print("Clustering data shape:", X_cluster.shape)
````

````python
inertia_values = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_cluster)
    inertia_values.append(kmeans.inertia_)

plt.figure(figsize=(8,5))

plt.plot(range(2, 11), inertia_values, marker='o')

plt.title('Elbow Method for Optimal Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')

plt.show()
````

````python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_cluster)

df_clean['Customer_Segment'] = clusters

print("Clustering completed successfully!")

print(df_clean['Customer_Segment'].value_counts().sort_index())
````
